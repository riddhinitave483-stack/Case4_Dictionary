```python

# Case Study 4: Dictionary Application using Hash Tables
# Collision Resolution Technique: Chaining

class Dictionary:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    # Hash function
    def hash_function(self, word):
        return sum(ord(char.lower()) for char in word) % self.size

    # Add word and meaning
    def insert(self, word, meaning):
        index = self.hash_function(word)

        # Update meaning if word already exists
        for item in self.table[index]:
            if item[0].lower() == word.lower():
                item[1] = meaning
                print("Word meaning updated successfully!")
                return

        # Add new word to the chain
        self.table[index].append([word, meaning])
        print("Word added successfully!")

    # Search for a word
    def search(self, word):
        index = self.hash_function(word)

        for item in self.table[index]:
            if item[0].lower() == word.lower():
                print("Word:", item[0])
                print("Meaning:", item[1])
                return

        print("Word not found in dictionary.")

    # Delete a word
    def delete(self, word):
        index = self.hash_function(word)

        for item in self.table[index]:
            if item[0].lower() == word.lower():
                self.table[index].remove(item)
                print("Word deleted successfully!")
                return

        print("Word not found in dictionary.")

    # Display all words
    def display(self):
        print("\n--- Dictionary Contents ---")
        found = False

        for index, chain in enumerate(self.table):
            if chain:
                found = True
                print("Index", index, end=": ")

                for word, meaning in chain:
                    print(f"[{word}: {meaning}]", end=" -> ")

                print("None")

        if not found:
            print("Dictionary is empty.")

    # Display hash table structure
    def show_table(self):
        print("\n--- Hash Table ---")

        for index, chain in enumerate(self.table):
            print(f"Index {index}: ", end="")

            if chain:
                print(" -> ".join(
                    f"({word}, {meaning})" for word, meaning in chain
                ))
            else:
                print("Empty")


# Main Program
dictionary = Dictionary()

while True:
    print("\n===== Dictionary Application =====")
    print("1. Add Word and Meaning")
    print("2. Search Word")
    print("3. Delete Word")
    print("4. Display All Words")
    print("5. Display Hash Table")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ").strip()

    if choice == "1":
        word = input("Enter word: ").strip()
        meaning = input("Enter meaning: ").strip()

        if word and meaning:
            dictionary.insert(word, meaning)
        else:
            print("Word and meaning cannot be empty.")

    elif choice == "2":
        word = input("Enter word to search: ").strip()

        if word:
            dictionary.search(word)
        else:
            print("Please enter a word.")

    elif choice == "3":
        word = input("Enter word to delete: ").strip()

        if word:
            dictionary.delete(word)
        else:
            print("Please enter a word.")

    elif choice == "4":
        dictionary.display()

    elif choice == "5":
        dictionary.show_table()

    elif choice == "6":
        print("Dictionary Application Closed.")
        break

    else:
        print("Invalid choice! Please enter 1 to 6.")
```
